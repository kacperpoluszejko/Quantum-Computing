```python
#EXERCISE 3.5
# I read that measurement in the Y-basis require HS' transformation, not HS'H
from IPython.display import display

print("x-basis")
qc_x = QuantumCircuit(1)

initial_state = [1, 0]
qc_x.initialize(initial_state, 0)

qc_x.h(0)

state = Statevector.from_instruction(qc_x)

print(state)
display(state.draw("latex"))

print("Y-basis")
qc_y = QuantumCircuit(1)

qc_y.initialize(initial_state, 0)

qc_y.sdg(0)
qc_y.h(0)

state_y = Statevector.from_instruction(qc_y)

print(state_y)
display(state_y.draw("latex"))

```

```python
#EXERCISE 3.6 a)

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display

print("2 qubit state: H|0> H|1>")

qc2 = QuantumCircuit(2)
qc2.x(1)
qc2.h(0)
qc2.h(1)

display(qc2.draw("mpl"))

state2 = Statevector.from_instruction(qc2)

display(state2.draw("latex", prefix="|\\Psi_2\\rangle = "))
display(state2.draw("qsphere"))
display(state2.draw("bloch"))


print("3 qubit state: H|1> H|0> H|0>")

qc3 = QuantumCircuit(3)
qc3.x(2)
qc3.h(0)
qc3.h(1)
qc3.h(2)

display(qc3.draw("mpl"))

state3 = Statevector.from_instruction(qc3)

display(state3.draw("latex", prefix="|\\Psi_3\\rangle = "))
display(state3.draw("qsphere"))
display(state3.draw("bloch"))


print("4 qubit state: H|0> H|0> H|0> H|1>")

qc4 = QuantumCircuit(4)
qc4.x(3)
qc4.h(0)
qc4.h(1)
qc4.h(2)
qc4.h(3)

display(qc4.draw("mpl"))

state4 = Statevector.from_instruction(qc4)

display(state4.draw("latex", prefix="|\\Psi_4\\rangle = "))
display(state4.draw("qsphere"))
display(state4.draw("bloch"))
```

```python
#Exercise 3.6 b)

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
from IPython.display import display

print("Bell state")

bell = QuantumCircuit(2)

bell.h(0)
bell.cx(0, 1)

display(bell.draw("mpl"))

bell_state = Statevector.from_instruction(bell)

display(bell_state.draw("latex", prefix="|\\Phi^+\\rangle = "))
display(bell_state.draw("qsphere"))
display(bell_state.draw("bloch"))


print("GHZ state")

ghz = QuantumCircuit(3)

ghz.h(0)
ghz.cx(0, 1)
ghz.cx(1, 2)

display(ghz.draw("mpl"))

ghz_state = Statevector.from_instruction(ghz)

display(ghz_state.draw("latex", prefix="|GHZ\\rangle = "))
display(ghz_state.draw("qsphere"))
display(ghz_state.draw("bloch"))
```

```python
#Exercise 3.8

from qiskit import QuantumCircuit
from qiskit.quantum_info import Operator
from IPython.display import display
import numpy as np

print("CNOT with control q0 and target q1")

qc1 = QuantumCircuit(2)
qc1.cx(0, 1)

display(qc1.draw("mpl"))

U1 = Operator(qc1).data

print(np.round(U1, 2))


print("\nCNOT with control q1 and target q0")

qc2 = QuantumCircuit(2)
qc2.cx(1, 0)

display(qc2.draw("mpl"))

U2 = Operator(qc2).data

print(np.round(U2, 2))
```
